```python
#MANUAL IMPLEMENTATION
import numpy as np
```

```python
def sigmoid(z):
    return 1/(1+ np.exp(-z))
```

```python
def sigmoid_derivative(z):
    return sigmoid(z) * (1 - sigmoid(z))
```

```python
X = np.array([[0.4, 0.2], [0.7, 0.8], [0.65, 0.86], [0.56, 0.73]])
y = np.array([[0], [1], [1], [1]])
```

```python
learning_rate = 0.1
```

```python
np.random.seed(42)
weights = np.random.rand(2, 1)
bias = np.random.rand(1)
print(f"Randomly initiates weights: {weights}")
print(f"Randomly initialize bias: {bias}")
```

```python
epochs = 10000
print(type(epochs))
```

```python
for epoch in range(epochs):
    linear_op = np.dot(X, weights) + bias
    y_predicted = sigmoid(linear_op)
    error = y - y_predicted
    der_pred_op = error*sigmoid_derivative(linear_op)
    weights +=learning_rate* np.dot(X.T, der_pred_op)
    bias +=learning_rate* np.sum(der_pred_op)
```

```python
print("Final Weights:", weights)
print("Final Bias", bias)
print("Predicted output:", y_predicted)
```

```python
#IMPLEMENTATION BY USING TENSORFLOW
import tensorflow as tf
inputs = np.array([[0,0], [0, 1], [1, 0], [1, 1]])
targets = np.array([[0], [1], [1], [1]])
```

```python
model = tf.keras.Sequential([tf.keras.layers.Dense(units=1, activation='sigmoid', input_shape=(2,))])
```

```python
model.compile(optimizer="full", loss="mean_squared_error")
```

```python
model.fit(inputs, targets, epochs=1000, verbose=0)
```

```python
print(model.predict(inputs))
```
